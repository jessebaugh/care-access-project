```python
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

# Load raw clinic data
clinics = pd.read_csv("../data/raw/clinics_raw.csv")

# Initialize geocoder
geolocator = Nominatim(user_agent="hiv_access_project")
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)

# Create address column if needed
if "full_address" not in clinics.columns:
    clinics["full_address"] = (
        clinics["address"] + ", " + clinics["city"] + ", " + clinics["state"]
    )

# Geocode
latitudes = []
longitudes = []

for addr in clinics["full_address"]:
    try:
        location = geocode(addr)
        if location:
            latitudes.append(location.latitude)
            longitudes.append(location.longitude)
        else:
            latitudes.append(None)
            longitudes.append(None)
    except:
        latitudes.append(None)
        longitudes.append(None)

clinics["latitude"] = latitudes
clinics["longitude"] = longitudes

# Save processed file
clinics.to_csv("../data/processed/clinics_geocoded.csv", index=False)

clinics.head()
```
